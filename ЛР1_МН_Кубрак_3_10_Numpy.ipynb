### Завдання 33
Нормалізуйте **кожен стовпець** матриці:

```text
A = [[10, 20, 30],
     [20, 30, 40],
     [30, 40, 50]]
```

за формулою:

\[
z = \frac{x - \mu}{\sigma}
\]

де `μ` — середнє значення стовпця, `σ` — стандартне відхилення стовпця.

Не використовуйте цикл `for`.

````python
# Код студента
A = np.array([[10, 20, 30],
              [20, 30, 40],
              [30, 40, 50]])

mu = A.mean(axis=0)
sigma = A.std(axis=0)

Z = (A - mu) / sigma

print("Матриця A:\n", A)
print("\nСереднє по стовпцях (mu):", mu)
print("Стандартне відхилення по стовпцях (sigma):", sigma)
print("\nНормалізована матриця Z:\n", Z)
````

```text
         [200, 180, 210, 230]]
```

Рядки — товари, стовпці — місяці.

Визначте:
1. загальний обсяг продажів кожного товару;
2. загальний обсяг продажів у кожному місяці;
3. товар із найбільшими сумарними продажами;
4. місяць із найбільшим загальним обсягом продажів;
5. середній обсяг продажів кожного товару.

````python
# Код студента
sales = np.array([[120, 150, 170, 160],
                   [80,  90,  100, 110],
                   [200, 180, 210, 230]])

total_per_product = sales.sum(axis=1)
total_per_month = sales.sum(axis=0)
best_product = np.argmax(total_per_product)
best_month = np.argmax(total_per_month)
mean_per_product = sales.mean(axis=1)

print("Загальний обсяг продажів кожного товару:", total_per_product)
print("Загальний обсяг продажів у кожному місяці:", total_per_month)
print("Товар із найбільшими продажами (індекс):", best_product)
print("Місяць із найбільшим обсягом продажів (індекс):", best_month)
print("Середній обсяг продажів кожного товару:", mean_per_product)
````

### Завдання 37
Створіть квадратну матрицю `5 × 5` із випадкових цілих чисел від `1` до `20`.

Виконайте:
1. знайдіть суму головної діагоналі;
2. знайдіть суму побічної діагоналі;
3. знайдіть максимальний елемент кожного рядка;
4. знайдіть мінімальний елемент кожного стовпця;
5. замініть усі елементи, менші за середнє значення всієї матриці, на `0`.

````python
# Код студента
np.random.seed(42)
A = np.random.randint(1, 21, size=(5, 5))

main_diag_sum = np.trace(A)
anti_diag_sum = np.trace(np.fliplr(A))
max_per_row = A.max(axis=1)
min_per_col = A.min(axis=0)

mean_val = A.mean()
B = np.where(A < mean_val, 0, A)

print("Матриця A:\n", A)
print("\nСума головної діагоналі:", main_diag_sum)
print("Сума побічної діагоналі:", anti_diag_sum)
print("Максимальний елемент кожного рядка:", max_per_row)
print("Мінімальний елемент кожного стовпця:", min_per_col)
print("\nСереднє значення матриці:", mean_val)
print("Матриця після заміни елементів < середнього на 0:\n", B)
````

### Завдання 40
Створіть матрицю коефіцієнтів і вектор правих частин для системи:

```text
 x + 2y +  z =  9
2x +  y -  z =  8
3x -  y + 2z =  3
```

1. Розв’яжіть систему за допомогою `np.linalg.solve()`.
2. Перевірте знайдений розв’язок матричним множенням `A @ x`.
3. Порівняйте результат із вектором `b`.

````python
# Код студента
A = np.array([[1, 2, 1],
              [2, 1, -1],
              [3, -1, 2]])

b = np.array([9, 8, 3])

x = np.linalg.solve(A, b)

check = A @ x

print("Розв'язок (x, y, z):", x)
print("Перевірка A @ x:", check)
print("Вектор b:", b)
print("Збігається:", np.allclose(check, b))
````
